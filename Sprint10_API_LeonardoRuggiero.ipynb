1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

```python
import requests
import pandas as pd

BASE_URL = "https://jsonplaceholder.typicode.com"
```

```python
res_posts = requests.get(f"{BASE_URL}/posts")
res_users = requests.get(f"{BASE_URL}/users")
res_todos = requests.get(f"{BASE_URL}/todos")

data_posts = res_posts.json()
data_users = res_users.json()
data_todos = res_todos.json()

print(f"Posts -> Cantidad total: {len(data_posts)} | Código de estado: {res_posts.status_code}")
print(f"Users -> Cantidad total: {len(data_users)} | Código de estado: {res_users.status_code}")
print(f"Todos -> Cantidad total: {len(data_todos)} | Código de estado: {res_todos.status_code}")
```

3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
res_not_found = requests.get(f"{BASE_URL}/posts/9999")

print(f"Código de estado recibido para un recurso inexistente: {res_not_found.status_code}")
```

4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
nueva_publicacion = {
    "title": "Mi primera publicación desde Python",
    "body": "Este es un contenido de prueba para la petición POST.",
    "userId": 1
}

res_post = requests.post(f"{BASE_URL}/posts", json=nueva_publicacion)

print(f"Código de estado (POST): {res_post.status_code}")
print("Respuesta JSON:")
print(res_post.json())
```


5. Fes una petició PATCH per modificar parcialment una publicació existent.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
datos_patch = {
    "title": "Título actualizado con PATCH"
}

res_patch = requests.patch(f"{BASE_URL}/posts/1", json=datos_patch)

print(f"Código de estado (PATCH): {res_patch.status_code}")
print("Respuesta JSON:")
print(res_patch.json())
```

6. Fes una petició DELETE sobre una publicació.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
res_delete = requests.delete(f"{BASE_URL}/posts/1")

print(f"Código de estado (DELETE): {res_delete.status_code}")
print("Respuesta JSON:")
print(res_delete.json())
```

## Nivell 3
API d'Open Data Barcelona



```python
BCN_API_URL = "https://opendata-ajuntament.barcelona.cat/data/api/action"
```

1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show. .

2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
url_busqueda = f"{BCN_API_URL}/package_search"
parametros_busqueda = {"q": "aparcaments", "rows": 10}

res_busqueda = requests.get(url_busqueda, params=parametros_busqueda)
busqueda_json = res_busqueda.json()

id_recurso = None

if busqueda_json.get("success"):
    datasets = busqueda_json["result"]["results"]
    print(f"Se han encontrado {len(datasets)} datasets relacionados.")

    for dataset in datasets:
        for recurso in dataset.get("resources", []):
            if recurso.get("datastore_active") is True:
                id_recurso = recurso["id"]
                print(f"¡Recurso con Datastore activo encontrado!")
                print(f"Dataset: {dataset['title']}")
                print(f"Nombre del recurso: {recurso['name']}")
                print(f"ID del recurso: {id_recurso}")
                break
        if id_recurso:
            break

if not id_recurso:
    print("\nNo se encontró ningún recurso dinámico con Datastore activo en la consulta.")
    print("Utilizando un ID de recurso con Datastore activo por defecto...")
    id_recurso = "1f2e624c-e83d-4c31-9257-897b7b252033"
```

3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres.

4. Converteix els resultats a un DataFrame.


5. Desa el DataFrame en un fitxer .csv.

```python
url_datastore = f"{BCN_API_URL}/datastore_search"
parametros_datastore = {
    "resource_id": id_recurso,
    "limit": 100
}

res_datastore = requests.get(url_datastore, params=parametros_datastore)
datastore_json = res_datastore.json()

if datastore_json.get("success"):
    registros = datastore_json["result"]["records"]

    df_bcn = pd.DataFrame(registros)
    print(f"¡DataFrame creado con éxito! Dimensiones: {df_bcn.shape}")

    display(df_bcn.head())

    nombre_archivo = "opendata_bcn_registros.csv"
    df_bcn.to_csv(nombre_archivo, index=False, encoding="utf-8")
    print(f"Archivo guardado correctamente como '{nombre_archivo}'.")
else:
    print("Error al obtener datos del datastore:", datastore_json.get("error"))
```
